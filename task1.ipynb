```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv('data.csv')

# 2. Display the first 5 rows
print("--- Dataset Preview ---")
display(df.head())

# 3. Check data types and missing values
print("\n--- Dataset Summary ---")
df.info()
```

```python
# 1. Count exact missing (null) values in every column
print("--- Missing Values Per Column ---")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0]) # Shows only columns with missing data

# 2. Check for duplicate rows
print("\n--- Total Duplicate Rows ---")
print(df.duplicated().sum())
```

```python
# 1. Remove the 5 duplicate rows
df = df.drop_duplicates()

# 2. Handle missing values in 'Region' (we can fill missing regions with 'Unknown' or drop them)
df['Region'] = df['Region'].fillna('Unknown')

# 3. Verify that cleaning worked (should show 0 nulls and 0 duplicates)
print("--- After Cleaning ---")
print("Remaining Missing Values:\n", df.isnull().sum())
print("Remaining Duplicates:", df.duplicated().sum())
print("New Dataset Shape (Rows, Columns):", df.shape)
```

```python
# Save the cleaned dataset as a new CSV file
df.to_csv('cleaned_dataset.csv', index=False)

print("Success! Cleaned dataset saved as 'cleaned_dataset.csv'")
```
